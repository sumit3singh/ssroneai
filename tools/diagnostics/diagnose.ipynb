```python
import subprocess
import os

print("--- Current Working Directory ---")
print(os.getcwd())

print("\n--- Running docker ps ---")
try:
    res = subprocess.run(["docker", "ps"], capture_output=True, text=True, check=True)
    print(res.stdout)
except Exception as e:
    print(f"Error: {e}")

print("\n--- Running docker-compose logs backend ---")
try:
    res = subprocess.run(["docker-compose", "logs", "--tail", "50", "backend"], capture_output=True, text=True, check=True)
    print(res.stdout)
except Exception as e:
    print(f"Error: {e}")
```

```python
import socket

def check_port(host, port):
    try:
        s = socket.create_connection((host, port), timeout=3)
        s.close()
        return True
    except Exception:
        return False

print("Port 5432 (PostgreSQL):", check_port("localhost", 5432))
print("Port 6379 (Redis):", check_port("localhost", 6379))
print("Port 8000 (FastAPI):", check_port("localhost", 8000))
```

```python
import sys
import os
import threading
import uvicorn

# Change working directory
os.chdir("e:/2026/baithak/backend")
sys.path.insert(0, os.getcwd())

# Force environment variables
os.environ["DB_HOST"] = "localhost"
os.environ["REDIS_HOST"] = "localhost"
os.environ["APP_SECRET_KEY"] = "baithak-super-secret-dev-key-change-in-production-2026"

def start_server():
    try:
        # Run on 0.0.0.0:8000
        uvicorn.run("src.app.server:app", host="0.0.0.0", port=8000, log_level="info")
    except Exception as e:
        print("Server error:", e)

# Start uvicorn thread
t = threading.Thread(target=start_server, daemon=True)
t.start()
print("FastAPI backend server thread started in background on port 8000!")
```
